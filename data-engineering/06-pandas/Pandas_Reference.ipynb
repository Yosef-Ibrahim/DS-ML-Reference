This interactive notebook demonstrates Pandas fundamentals: DataFrames, filtering, vectorized updates, groupby aggregations, handling missing data, and datetime filtering.

```python
import pandas as pd
import numpy as np

# Create sample DataFrame
data = {
    'ID': [1, 2, 3, 4, 5],
    'Name': ['John Doe', 'Jane Smith', 'Bob Johnson', 'Alice White', 'Charlie Brown'],
    'Department': ['HR', 'IT', 'Finance', 'IT', 'Marketing'],
    'Age': [28, 34, 45, 29, 32],
    'Salary': [50000, 75000, 85000, 72000, 68000],
    'Joining_Date': ['2020-05-21', '2019-06-12', '2018-07-15', '2021-01-04', '2019-11-30']
}
df = pd.DataFrame(data)
df.head()
```

## 1. Filtering & Vectorized Operations

```python
# Filter IT Department Employees
it_employees = df[df['Department'] == 'IT']
it_employees
```

```python
# Vectorized column addition (5% raise)
df['Salary_After_Raise'] = df['Salary'] * 1.05
df[['Name', 'Salary', 'Salary_After_Raise']]
```

## 2. Groupby Aggregations & Missing Data

```python
# Average salary by department
df.groupby('Department')['Salary'].mean().reset_index()
```

```python
# Handling missing values (fillna)
df_nan = pd.DataFrame({
    'A': [1.0, 2.0, np.nan],
    'B': [4.0, np.nan, 6.0]
})
df_nan.fillna(0)
```
