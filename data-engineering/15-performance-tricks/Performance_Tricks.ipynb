This notebook benchmarks a bounded, local SQLite workload with only the Python standard library. It demonstrates measuring a baseline before changing a query.

```python
import sqlite3
import time

row_count = 20_000
connection = sqlite3.connect(':memory:')
connection.execute('CREATE TABLE events (event_id INTEGER PRIMARY KEY, region TEXT, amount REAL)')
connection.executemany(
    'INSERT INTO events VALUES (?, ?, ?)',
    ((i, f'region-{i % 10}', float(i % 100)) for i in range(row_count)),
)
connection.commit()

start = time.perf_counter()
result = connection.execute(
    'SELECT region, SUM(amount) FROM events WHERE event_id >= ? GROUP BY region',
    (row_count // 2,),
).fetchall()
elapsed = time.perf_counter() - start
print('rows:', len(result), 'elapsed_seconds:', round(elapsed, 6))
```
