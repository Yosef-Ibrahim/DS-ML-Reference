This interactive notebook demonstrates basic Matplotlib plots (Line, Scatter, Subplots) and Seaborn statistical plots (Boxplots, Violin Plots, Histograms, Correlation Heatmaps).

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np

sns.set_theme(style="whitegrid")

# Create synthetic data
np.random.seed(42)
df = pd.DataFrame({
    'total_bill': np.random.normal(loc=20, scale=5, size=100),
    'tip': np.random.normal(loc=3, scale=1, size=100),
    'category': np.random.choice(['Group A', 'Group B'], size=100)
})
df.head()
```

## 1. Matplotlib Subplots

```python
fig, axes = plt.subplots(1, 2, figsize=(10, 4))
axes[0].scatter(df['total_bill'], df['tip'], color='blue', alpha=0.7)
axes[0].set_title("Total Bill vs Tip")

axes[1].hist(df['total_bill'], bins=15, color='green', edgecolor='black')
axes[1].set_title("Total Bill Distribution")
plt.tight_layout()
plt.show()
```

## 2. Seaborn Boxplots & Correlation Heatmap

```python
plt.figure(figsize=(6, 4))
sns.boxplot(x='category', y='total_bill', data=df)
plt.title("Total Bill by Category")
plt.show()
```

```python
plt.figure(figsize=(6, 4))
corr = df[['total_bill', 'tip']].corr()
sns.heatmap(corr, annot=True, cmap='coolwarm', vmin=-1, vmax=1)
plt.title("Correlation Matrix")
plt.show()
```
