## 1. String Methods & Formatting

```python
first_name = "Youssef"
last_name = "Ibrahim"
full_name = f"{first_name} {last_name}"

print(f"Full Name: {full_name}")
print(f"Uppercase: {full_name.upper()}")
print(f"Lowercase: {full_name.lower()}")
print("Escaped Tab & Newline:\n\tData Engineering\n\tPython Reference")
```

## 2. Numbers & Operators

```python
print(f"Addition (2 + 3): {2 + 3}")
print(f"Exponentiation (2 ** 3): {2 ** 3}")
print(f"Float Division (7 / 2): {7 / 2}")
print(f"Floor Division (7 // 2): {7 // 2}")
print(f"Modulo (7 % 2): {7 % 2}")
```

## 3. Lists & Manipulations

```python
motorcycles = ['honda', 'yamaha', 'suzuki']
motorcycles.append('ducati')
motorcycles.insert(0, 'bmw')
print(f"Modified List: {motorcycles}")

popped_item = motorcycles.pop()
print(f"Popped Item: {popped_item}")
print(f"List After Pop: {motorcycles}")

motorcycles.sort()
print(f"Sorted List: {motorcycles}")
```

## 4. Range Generators & List Comprehensions

```python
numbers = list(range(1, 11))
print(f"Numbers 1-10: {numbers}")
print(f"Min: {min(numbers)}, Max: {max(numbers)}, Sum: {sum(numbers)}")

squares = [val**2 for val in range(1, 11)]
print(f"Squares: {squares}")

even_squares = [val**2 for val in range(1, 11) if val % 2 == 0]
print(f"Even Squares: {even_squares}")
```
