This notebook demonstrates an MLflow-shaped experiment record using only the Python standard library. It runs locally without credentials or an MLflow installation.

```python
import json
import math
from dataclasses import asdict, dataclass

@dataclass
class Run:
    experiment: str
    parameters: dict
    metrics: dict
    tags: dict

observations = [(1.0, 2.1), (2.0, 4.0), (3.0, 6.2), (4.0, 8.1)]
slope = sum(x * y for x, y in observations) / sum(x * x for x, _ in observations)
rmse = math.sqrt(sum((y - slope * x) ** 2 for x, y in observations) / len(observations))
run = Run('notebook_regression', {'seed': 7}, {'rmse': round(rmse, 6)}, {'stage': 'candidate'})
print(json.dumps(asdict(run), indent=2))
```

In a production workshop, replace the local record with `mlflow.start_run()`, `log_params`, `log_metrics`, and `log_artifact`, then validate the model signature before promotion.
