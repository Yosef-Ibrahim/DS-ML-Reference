This interactive notebook demonstrates Advanced Pandas: Concatenation, Relational Merging, GroupBy Transform & Filter, Pivot Tables, and Time Series Resampling.

```python
import pandas as pd
import numpy as np

# 1. Concatenation
df1 = pd.DataFrame({'A': ['A0', 'A1'], 'B': ['B0', 'B1']})
df2 = pd.DataFrame({'A': ['A2', 'A3'], 'B': ['B2', 'B3']})
pd.concat([df1, df2], ignore_index=True)
```

```python
# 2. Relational Merging
df_emp = pd.DataFrame({'emp_id': [1, 2, 3], 'name': ['Alice', 'Bob', 'Charlie'], 'dept_id': [10, 20, 10]})
df_dept = pd.DataFrame({'department_id': [10, 20, 30], 'dept_name': ['IT', 'HR', 'Finance']})
pd.merge(df_emp, df_dept, left_on='dept_id', right_on='department_id', how='inner')
```

```python
# 3. Transform & Filter
df_sales = pd.DataFrame({
    'Store': ['A', 'A', 'B', 'B', 'A'],
    'Sales': [100, 200, 150, 300, 120]
})
df_sales['Centered_Sales'] = df_sales.groupby('Store')['Sales'].transform(lambda x: x - x.mean())
df_sales
```

```python
# 4. Pivot Tables
df_titanic = pd.DataFrame({
    'Sex': ['female', 'female', 'male', 'male', 'female'],
    'Class': ['First', 'Second', 'First', 'Third', 'First'],
    'Fare': [100.0, 50.0, 110.0, 20.0, 120.0],
    'Embarked': ['Southampton', 'Cherbourg', 'Southampton', 'Southampton', 'Cherbourg']
})
df_titanic.pivot_table(values='Fare', index=['Sex', 'Class'], columns='Embarked', aggfunc='mean', fill_value=0)
```
