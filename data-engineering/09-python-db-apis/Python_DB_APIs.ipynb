This interactive notebook demonstrates SQL database connections with SQLite, data ingestion logging, and consuming external REST APIs.

```python
import sqlite3
import pandas as pd

# Connect to in-memory SQLite database
conn = sqlite3.connect(':memory:')
cursor = conn.cursor()

cursor.execute('''
CREATE TABLE pipeline_logs (
    log_id INTEGER PRIMARY KEY AUTOINCREMENT,
    pipeline_name TEXT NOT NULL,
    status TEXT NOT NULL,
    records_processed INTEGER
);
''')

logs = [
    ('batch_etl_ecommerce', 'SUCCESS', 15000),
    ('realtime_fraud_stream', 'SUCCESS', 84000),
    ('cloud_lakehouse_sync', 'FAILED', 0)
]

cursor.executemany('''
INSERT INTO pipeline_logs (pipeline_name, status, records_processed)
VALUES (?, ?, ?);
''', logs)
conn.commit()

pd.read_sql_query('SELECT * FROM pipeline_logs;', conn)
```
