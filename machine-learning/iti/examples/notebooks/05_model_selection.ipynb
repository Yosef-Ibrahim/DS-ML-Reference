# Cross-validation and grid search

Keep scaling inside the pipeline so every fold is isolated.


```python
from sklearn.datasets import load_iris
from sklearn.model_selection import GridSearchCV
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
data = load_iris()
pipeline = Pipeline([('scale', StandardScaler()), ('model', LogisticRegression(max_iter=1000))])
search = GridSearchCV(pipeline, {'model__C': [0.1, 1, 10]}, cv=5)
search.fit(data.data, data.target)
print(search.best_params_, search.best_score_)
```
