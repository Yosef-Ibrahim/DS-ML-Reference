# Unsupervised learning addition

Scale features before K-means and inspect a PCA projection.


```python
from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
from sklearn.metrics import silhouette_score
X, _ = load_iris(return_X_y=True)
Xs = StandardScaler().fit_transform(X)
labels = KMeans(n_clusters=3, n_init=20, random_state=42).fit_predict(Xs)
Z = PCA(n_components=2).fit_transform(Xs)
print(silhouette_score(Xs, labels), Z.shape)
```
