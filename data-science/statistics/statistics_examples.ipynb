# Statistics ITI examples
IQR, sample spread, z-scores, correlation, and least-squares regression.

```python
from math import sqrt
values = [2, 3, 4, 5, 6, 20]
ordered = sorted(values)
median = lambda a: a[len(a)//2] if len(a)%2 else (a[len(a)//2-1] + a[len(a)//2]) / 2
q1, q3 = median(ordered[:3]), median(ordered[3:])
print('median:', median(ordered), 'IQR:', q3-q1, 'upper fence:', q3 + 1.5*(q3-q1))
```

```python
x, y = [1, 2, 3, 4, 5], [2, 4, 5, 8, 10]
xm, ym = sum(x)/len(x), sum(y)/len(y)
slope = sum((a-xm)*(b-ym) for a,b in zip(x,y)) / sum((a-xm)**2 for a in x)
intercept = ym - slope*xm
r = sum((a-xm)*(b-ym) for a,b in zip(x,y)) / sqrt(sum((a-xm)**2 for a in x)*sum((b-ym)**2 for b in y))
print('r:', round(r, 3), 'line:', round(intercept, 3), '+', round(slope, 3), '* x')
```
