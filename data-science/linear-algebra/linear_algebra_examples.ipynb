# Linear Algebra ITI examples
Standard-library dot product, system solving, and rotation.

```python
from math import acos, cos, degrees, pi, sin, sqrt
u, v = (3, 4), (4, 0)
dot = sum(a * b for a, b in zip(u, v))
print(dot, degrees(acos(dot / (sqrt(25) * sqrt(16)))))
```

```python
def solve_2x2(A, b):
    (a, c), (d, e) = A
    det = a * e - c * d
    if det == 0: raise ValueError('singular matrix')
    return ((b[0] * e - c * b[1]) / det, (a * b[1] - b[0] * d) / det)
A, b = ((2, 1), (1, -1)), (7, 1)
x = solve_2x2(A, b)
print(x, [sum(a * z for a, z in zip(row, x)) for row in A])
```
